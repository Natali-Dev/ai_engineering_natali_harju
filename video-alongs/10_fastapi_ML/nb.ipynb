```python
from pydantic import BaseModel, Field

class Iris(BaseModel): 
    sepal_length: float  = Field(gt=4,lt=8.5)
    sepal_width: float = Field(gt=1.8,lt=5)
    petal_length: float = Field(gt=0.8,lt=7.5)
    petal_width: float = Field(gt=0,lt=3)
```

```python
payload = Iris(sepal_length=4.2, sepal_width=2.5, petal_length=0.9, petal_width=1)

payload.model_dump()
```

```python
import pandas as pd
pd.DataFrame([payload.model_dump()])
```

```python
import httpx
import requests
payload = {
    "sepal_length": 5,
    "sepal_width": 4.5,
    "petal_length": 1.8,
    "petal_width": 1
}

url = "http://127.0.0.1:8000/api/iris/predict"

with httpx.Client(timeout=10) as client:
    response = client.post(url, json= payload)
    response.raise_for_status()
response.json()
# response = requests.get(url)
# response.text
```
