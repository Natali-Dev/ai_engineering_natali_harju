## 4. Theory questions

a) What is a REST API?
- En arkitektur för att bygga API-program. Använder CRUD. 
b) What is CRUD?
- Create, Read, Update, Delete. Används med API/ Kallas för CRUD-APP, en modell/ett tankesätt när man bygger en API-hemsida

c) Why would you want to use an API, can't you just create everything in for example a streamlit app?
- API är användbart för att communicate med backend och frontend. Du vill inte hämta datan från källan till frontend för att: Det blir osäkert och svårhanterligt. Med ett API-lager kan datan valideras och åtkomst kontrolleras först! 

d) What does it mean with decoupling and tightly coupling?
- tightly coupling: Att hämta data till din frontend dirket från källan
- decoupling: Att ha ett API lager emellan din front och backend, vilket är bättre
e) What are some other frameworks to create APIs in Python?
- Flask, Django REST Framework, Tornado, Falcon, Pyramid
f) How would you mix Python together with a JavaScript frontend?
- python i backend, javascript i frontend, med ett api-lager emellan.

| terminology     | explanation |
| --------------- | ----------- |
| uvicorn         | det du använder för att köra din API-sida, uvicorn api:app            |
| endpoint        | slutet på din URL, tex /docs, /Api/Sales            |
| path parameter  | En variabel i din URL, {id}            |
| query parameter | Parametrar efter ? i URL. Importera Query och använd denna för att lägga till, Används exakt som Field i som argument            |
| routes          | används istället för @app för att ha en specifik endpoint. Definerar endpoints             |
| request         | en modul du importerar, använd med .get för att göra ett HTTP anrop. Alla HTTP-anorp är inte API-anrop             |
| REST            | Arkitekturstil för API:er, Representational State Transfer, använder CRUD            |
| CRUD            | Create, Read, Update, Delete             |
| put             | Skapar eller ersätter data            |
| post            | Skapa ny data            |
| read            | @app.get - läser av datan            |
| update          | @app.post - ändrar i datan, kan vara put också.            |
| field_validator | pydantic-decorator, för att validera fält i en modell (klass)            |
| Field           | En metod från pydantic där man kan lägga in tex. gt, lt till fält i klassen            |
| response        | Vad API returnar. ta .text för att läsa vad den innehåller            |
| swagger ui      | UI för uvicorn, lägg till /docs för att öppna            |
| OpenAPI         | Standard för att beskriva REST API:er (JSON/YAML-specifikation). FastAPI genererar detta automatiskt            |
| curl            | kommando för att göra HTTP-anrop            |
