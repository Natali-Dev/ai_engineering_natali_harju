```python
from pydantic import BaseModel
from pydantic import ValidationError
from pydantic import Field
# a) Create a BaseModel for a User. It should have a required id (integer) and a required name (string). 
# Instantiate the model with valid data and then with invalid data (e.g., a string for id) to see the ValidationError.
class User(BaseModel): 
    id: int = Field (lt= 100)
    name: str
print(User(id=1, name="Axel"))
try: 
    User(id=101, name="Axel")
except ValidationError as err: 
    print(err)
```

```python

# c) Use normal python class to replicate what you have created in b), i.e. create a Person class with proper input validation.

class Person:
    def __init__(self, name: str, age: int, email: str, favorite_pet: str):
        # if not isinstance(name, str):
        #     raise TypeError(f"Name must be of type str, not {type(name)}")

        # if not isinstance(email, str):
        #     raise TypeError(f"email must be of type str, not {type(email)}")
        # if not isinstance(favorite_pet, str):
        #     raise TypeError(f"pet must be of type str, not {type(favorite_pet)}")        
        self.name = name
        self.age = age
        self.email = email
        self.favorite_pet = favorite_pet

    # def email(self): 
    #     return self._email
    # def favorit_pet(self): 
    #     return self._favorit_pet
    @property #getter
    def name(self): 
        return self._name
    
    @name.setter #setter
    def name(self, value: str):
            if not isinstance(value, str):
                raise TypeError(f"name must be of type str, not {type(value)}")
            self._name = value
            
    @property
    def age(self): 
        return self._age
    
    @age.setter #setter
    def age(self, value: int):
            if not isinstance(value, int):
                raise TypeError(f"age must be of type int, not {type(value)}")
            self._age = value

    @property
    def email(self): 
        return self._email
    
    @email.setter #setter
    def email(self, value: str):
            if not isinstance(value, str):
                raise TypeError(f"email must be of type int, not {type(value)}")
            elif not value.__contains__('@' and '.'):
                raise ValueError(f"email must contain '@' and ' . ' ")
            elif value.__contains__(' '):
                raise ValueError(f"email must not contain spaces! ")
            self._email = value

    @property #getter
    def favorite_pet(self): 
        return self._favorite_pet
    
    @favorite_pet.setter #setter
    def favorite_pet(self, value: str):
            if not isinstance(value, str):
                raise TypeError(f"favorite pet must be of type str, not {type(value)}")
            if value == "Esko" or value == "Dog": # or not value == "Dog":
                self._favorite_pet = value
            else: 
                raise ValueError(f"Only allowed type of pet is either Dog or Esko!")

try:
    person = Person(name="Natali", age=20, email="Hej@gmai.com", favorite_pet="Dog")
    print(person.name, person.age, person.email, person.favorite_pet)
except ValueError as err: 
    print(err)
```
