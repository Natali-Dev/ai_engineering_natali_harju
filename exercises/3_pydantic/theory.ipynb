## 4. Theory questions

a) Explain the primary purpose of Pydantic. What core problems does it solve for developers?
- Används främst för variabel-validering (datavalidering och datamodellering), och man kan ange väldigt specifika parametrar med tex Field, Enum och Literal. Ungefär som C# gjorde fast här kan man importera mer saker där du då kan ange specifika parametrar för varje variabel. Bra när man vill att inkommande data ska ha en specifik struktur. Ger också tydliga felmeddelanden när data inte matchar modellen. 

b) In Pydantic, what is the key difference between using model_validate() and model_validate_json()? When would you use each one?
- model_validate = för att validera ett python-objekt (list, dict, list[dict])
- model_validate_json = för att validera en JSON-string

c) How can you define a field in a Pydantic model that has a specific value constraint, such as a minimum or maximum number? Give an example.
- grades: Literal["A", "B"]
- age: int = Field(gt=-1, lt=126) #värdet får alltså vara 0 upp till max 125. 

d) Describe how Pydantic's data validation can improve the robustness and reliability of an application that processes data from an external API.
- Om age skulle komma som en string "30" validerar den ändå som en int. Kanske vill man ha variabeln i ett exakt format, då kan inte data som man inte vill ha följa med. Säg att vi har tränat en modell utan outliers, och ska fortsätta predicta på data enbart inom ett visst omfång, då blir det väldigt smidigt med pydantic. 

e) What is a computed field in Pydantic?
- Ett fält som inte lagras direkt i modellen. Du kan tex räkna antal ord i en str eller hur många output listan innehåller. Deklareras med dekoratorn @computed_field

f) What are tokens in LLMs?
- hur många textenheter du kan skicka in till den för att den ska skicka ut ett bra svar, om du överskrider token-mängden på 32.000 hos gemini (ca 915 rader i en .txt, ink. tidigare konversationer.) så måste den ta genvägar, komprimera din inskickade text och ger inte ett lika bra svar.
- det är ungefär 35 tokens/rad i en .txt.

g) Large language models (LLMs) can struggle to generate a large amount of structured data, like 100 JSON objects in a single request, primarily due to token limitations and context window constraints.
- Stämmer, och det är inte bara pga. token-gränser. En annan orsak är LLM:s tendens att tappa struktur och "halliciunera". Contex window = det minne språkmodellen har i en konversation. Om du överskrider token-mängden kommer den excludera tidigare konversationer, det blir ett mindre fönster (?).

h) What are some strategies you can use to make an LLM's output more reliable when you need it to produce a specific format, such as JSON or Markdown?
- Att ha tydliga regler och exempel helt enkelt. 

i) What are few shot examples?
- lägga till "a few shot examples", det är att tydligt skriva output i det format du vill ha det med tex. måsvingar och list-symboler, ge den regler på hur den ska bete sig. Du kan också skicka meningar som exempel och be den följa det språkmönstret, då kommer den följa samma struktur på sin output baserat på det.
