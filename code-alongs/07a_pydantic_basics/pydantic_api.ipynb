## Use pydantic to validate API data and work with it in an OOP manner

```python
import requests

URL = "https://pokeapi.co/api/v2/pokemon?limit=20"
data = requests.get(URL).json()
data
```

## Deserialization - turn json-object into a pydantic model
- data has a key called results.
    - inside result there is a list of dictionarys
- count gives us the total amout of pokemons availible to get from url 

```python
from pydantic import BaseModel

class PokemonList(BaseModel):
    results: list[dict]
    count: int

pokemons = PokemonList.model_validate(data)
pokemons.count
```

```python
pokemons.results
```

```python
from pydantic import computed_field

class PokemonList(BaseModel):
    results: list[dict]
    count: int
    
    @computed_field # decorator. beräknar längden på results och lägger in det som ett extra fält i objektet
    @property
    def number_stored_pokemon(self) -> int: # ej typehinting, int blir mandatory pga pydantic
        return len(self.results)
    
pokemons = PokemonList.model_validate(data)
pokemons.number_stored_pokemon
```

```python
pokemons.name
```

## Serialize into json - konvertera till json-objekt från pydantic model 

```python
with open("pokemons.json", "w") as file: 
    file.write(pokemons.model_dump_json(indent=3))
```
