```python
class Person:
    def __init__(self, name: str, gender: str, age: int):
        if not isinstance(name, str):
            raise TypeError(f"name must be f type str not {type(name)}")
        self.name = name
        self.gender = gender
        self.age = age
        
    @property #getter, returnar _age 
    def age(self): 
        return self._age
    
    @age.setter
    def age(self, value: int):
        if not isinstance (value, int): 
            raise TypeError(f"age must be f type str not {type(value)} that you provided")
        if value < 0 or value > 125: 
            raise ValueError(f"age must be between 0-125, not {value} that you provided")
        
        self._age = value
try:
    person2 = Person(name="Natali", age= 140, gender="F")
except ValueError as err: 
    print(err)
# person2.age
# try: 
#     Person(name=12, gender="F", age="27")
# except TypeError as err: 
#     print(err)
```

## work smarter with pydantic

```python
from pydantic import BaseModel

#inhertit from basemodel -> makes it a pydantic model, but still a normal class
class Person(BaseModel): 
    name: str
    age: int
    gender: str

Person(name="1", gender="f", age= "32")
```

```python
from pydantic import ValidationError
try:
    Person(name=1, gender=True, age= 230)
except ValidationError as err: 
    print(err)
```

### bad, we have negative age

```python
try:
    print(Person(name= "Sofia", gender="F", age= -23))
except ValidationError as err: 
    print(err)
```

## Fix person class with age validation

```python
from pydantic import Field
class Person(BaseModel): 
    name: str
    age: int = Field(gt=-1, lt=126) #gt = Greater Than, lt = Lower Than
    gender: str
try:
    Person(name="Kim", gender="M", age= 500)
except ValidationError as err: 
    print(err)
```

### Make the instance into a dictionary

```python

person3 = Person(name="Kim", gender="M", age= 50)
person3.model_dump() # Få ut som en dictionary
```

### Make the instance into a json string

```python
person3.model_dump_json() # få ut som json-string - serialization
```
